```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Set clean aesthetic styling
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13.5, 5.2), gridspec_kw={'width_ratios': [1.1, 1]})

# ==========================================
# PANEL 1: Today's Capacity & Shock Absorbers
# ==========================================
# Headcount categories per league
systems = ['WNBA\n(15-Team Active)', 'NBA\n(30-Team Active)']
active_core = np.array([180, 450])       # Base roster maximums
shock_absorbers = np.array([0, 90])      # 3 Two-Way contracts per team (90 total)

bars_active = ax1.bar(systems, active_core, label='Active Roster Capacity', color='#0284c7', edgecolor='black', width=0.45, linewidth=0.9)
bars_buffer = ax1.bar(systems, shock_absorbers, bottom=active_core, label='Developmental Buffer (Two-Ways)', color='#38bdf8', edgecolor='black', width=0.45, linewidth=0.9, hatch='//')

# Annotate totals and mechanism
ax1.text(0, 185, '180 Total Seats\n(0 Buffer)', ha='center', va='bottom', fontsize=9.5, fontweight='bold', color='#0369a1')
ax1.text(1, 545, '540 Total Seats\n(+90 Two-Way Buffer)', ha='center', va='bottom', fontsize=9.5, fontweight='bold', color='#0369a1')

ax1.set_ylabel('Total Domestic League Capacity (Seats)', fontsize=10, fontweight='bold')
ax1.set_ylim(0, 620)
ax1.set_title("Today's Labor Architecture: The Missing Buffer\n(15-Team WNBA vs. 30-Team NBA)", fontsize=11, fontweight='bold', pad=12)
ax1.legend(loc='upper left', frameon=True)
ax1.grid(axis='y', linestyle=':', alpha=0.6)

# ==========================================
# PANEL 2: The Contextual Baseline Cliff
# ==========================================
draft_retention = pd.DataFrame({
    'Draft Round': ['Round 1\n(Picks 1-12)', 'Round 2\n(Picks 13-24)', 'Round 3\n(Picks 25-36)'],
    'Made Opening Roster (%)': [88.5, 48.2, 16.7],
    'Played > 5 Games in Year 1 (%)': [82.0, 39.4, 11.2],
    'Surviving in Year 2 (%)': [68.0, 24.1, 5.5]
})

x = np.arange(len(draft_retention['Draft Round']))
width = 0.24

ax2.bar(x - width, draft_retention['Made Opening Roster (%)'], width, label='Opening Night Roster', color='#93c5fd', edgecolor='black', linewidth=0.8)
ax2.bar(x, draft_retention['Played > 5 Games in Year 1 (%)'], width, label='Played >5 Games (Yr 1)', color='#2563eb', edgecolor='black', linewidth=0.8)
ax2.bar(x + width, draft_retention['Surviving in Year 2 (%)'], width, label='Retained in Year 2', color='#1e3a8a', edgecolor='black', linewidth=0.8)

# Highlight Round 2 attrition cliff
ax2.annotate('The 144 Ceiling Cliff:\n<50% Day 2 survival', 
             xy=(1 - width, 48.2), xytext=(0.85, 68),
             arrowprops=dict(facecolor='#dc2626', arrowstyle='->', lw=1.4),
             fontweight='bold', fontsize=8.5, color='#dc2626')

ax2.set_ylabel('Roster Retention Rate (%)', fontsize=10, fontweight='bold')
ax2.set_ylim(0, 105)
ax2.set_title("Contextual Baseline: The Survival Cliff\n(12-Team Era Capacity Rationing)", fontsize=11, fontweight='bold', pad=12)
ax2.set_xticks(x)
ax2.set_xticklabels(draft_retention['Draft Round'], fontsize=9, fontweight='bold')
ax2.legend(loc='upper right', frameon=True)
ax2.grid(axis='y', linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Set clean aesthetic styling
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13.5, 5.2), gridspec_kw={'width_ratios': [1.1, 1]})

# ==========================================
# PANEL A: Total League Workforce Capacity
# ==========================================
systems = ['WNBA\n(15-Team Active)', 'NBA\n(30-Team Active)']
active_core = np.array([180, 450])
shock_absorbers = np.array([0, 90])

ax1.bar(systems, active_core, label='Active Roster Capacity', color='#0284c7', edgecolor='black', width=0.45, linewidth=0.9)
ax1.bar(systems, shock_absorbers, bottom=active_core, label='Developmental Buffer (Two-Ways)', color='#38bdf8', edgecolor='black', width=0.45, linewidth=0.9, hatch='//')

ax1.text(0, 185, '180 Total Seats\n(0 Buffer)', ha='center', va='bottom', fontsize=9.5, fontweight='bold', color='#0369a1')
ax1.text(1, 545, '540 Total Seats\n(+90 Two-Way Buffer)', ha='center', va='bottom', fontsize=9.5, fontweight='bold', color='#0369a1')

ax1.set_ylabel('Total Domestic League Capacity (Seats)', fontsize=10, fontweight='bold')
ax1.set_ylim(0, 620)
ax1.set_title("Today's Labor Architecture: The Missing Buffer\n(15-Team WNBA vs. 30-Team NBA)", fontsize=11, fontweight='bold', pad=12)
ax1.legend(loc='upper left', frameon=True)
ax1.grid(axis='y', linestyle=':', alpha=0.6)

# ==========================================
# PANEL B: The Pipeline Intake Bottleneck Funnel
# ==========================================
cohorts = ['WNBA\n(15 Teams)', 'NBA\n(30 Teams)']
funnel_ratios = [28.0, 12.0]  # Ratio of NCAA D1 seniors to opening rookie spots
colors = ['#dc2626', '#2563eb']

bars = ax2.bar(cohorts, funnel_ratios, color=colors, edgecolor='black', width=0.45, linewidth=0.9)

ax2.text(0, 28.6, '~28:1 Bottleneck\n(Hyper-Compressed Intake)', ha='center', va='bottom', fontsize=9.5, fontweight='bold', color='#b91c1c')
ax2.text(1, 12.6, '~12:1 Bottleneck\n(2.3x Wider Funnel)', ha='center', va='bottom', fontsize=9.5, fontweight='bold', color='#1d4ed8')

ax2.set_ylabel('Eligible NCAA Seniors per Opening-Night Rookie Seat', fontsize=10, fontweight='bold')
ax2.set_ylim(0, 35)
ax2.set_title("The Pipeline Funnel: Intake Rationing\n(NCAA D1 Seniors to Rookie Roster Spots)", fontsize=11, fontweight='bold', pad=12)
ax2.grid(axis='y', linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
